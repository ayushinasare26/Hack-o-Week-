```python
A = np.array([
    [2, 1],
    [1, 2]
])

vectors = np.array([
    [1, 0],
    [0, 1]
])

transformed_vectors = A @ vectors.T

plt.figure(figsize=(7, 7))

# Original vectors
for vector in vectors:
    plt.quiver(
        0, 0,
        vector[0],
        vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        label='Original'
    )

# Transformed vectors
for vector in transformed_vectors.T:
    plt.quiver(
        0, 0,
        vector[0],
        vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        label='Transformed'
    )

plt.xlim(-1, 4)
plt.ylim(-1, 4)
plt.axhline(0)
plt.axvline(0)
plt.grid()

plt.title("Matrix Transformation")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = 5
learning_rate = 0.1

x_history = []
y_history = []

for i in range(20):

    x_history.append(x)
    y_history.append(x**2)

    gradient = 2 * x

    x = x - learning_rate * gradient


# Function curve
x_values = np.linspace(-5, 5, 200)
y_values = x_values**2

plt.figure(figsize=(9, 6))

plt.plot(
    x_values,
    y_values,
    label="f(x) = x²"
)

plt.scatter(
    x_history,
    y_history,
    label="Gradient Descent Steps"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Finding the Minimum")

plt.legend()
plt.grid()

plt.show()
```

```python
def chain_rule_derivative(x):

    u = 2*x + 1

    dy_du = 2*u
    du_dx = 2

    dy_dx = dy_du * du_dx

    return dy_dx


x = 3

print("x =", x)
print("Derivative using Chain Rule =", chain_rule_derivative(x))
```

```python
def function(x):
    return (2*x + 1)**2


x = 3
h = 0.00001

numerical_derivative = (
    function(x + h) - function(x - h)
) / (2*h)

analytical_derivative = chain_rule_derivative(x)

print("Numerical Derivative:", numerical_derivative)
print("Analytical Derivative:", analytical_derivative)
```

```python
# Input
x = 2

# Weight
w = 3

# Bias
b = 1

# Forward propagation
z = w*x + b

# Prediction
y_pred = z

# Actual value
y_true = 10

# Error
error = (y_pred - y_true)**2

print("Input:", x)
print("Weight:", w)
print("Bias:", b)
print("Prediction:", y_pred)
print("Actual:", y_true)
print("Error:", error)
```

```python
x = 2
w = 3
b = 1
y_true = 10

# Forward pass
z = w*x + b
y_pred = z

# Calculate gradients
dLoss_dz = 2 * (y_pred - y_true)
dz_dw = x

dLoss_dw = dLoss_dz * dz_dw

print("Prediction:", y_pred)
print("Loss Gradient:", dLoss_dz)
print("Weight Gradient:", dLoss_dw)
```

```python
x = 2
w = 3
b = 1

y_true = 10

learning_rate = 0.01

loss_history = []

for epoch in range(50):

    # Forward pass
    y_pred = w*x + b

    # Calculate loss
    loss = (y_pred - y_true)**2

    loss_history.append(loss)

    # Backpropagation
    dLoss_dy = 2 * (y_pred - y_true)
    dy_dw = x

    gradient = dLoss_dy * dy_dw

    # Update weight
    w = w - learning_rate * gradient

print("Final Weight:", w)
print("Final Prediction:", w*x + b)
print("Final Loss:", loss_history[-1])
```

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss During Training")

plt.grid()

plt.show()
```
